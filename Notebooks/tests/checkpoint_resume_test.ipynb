# Checkpoint Resume Test (Test 1)

This notebook validates that checkpointing works correctly.

Procedure:

1. Run training with a small configuration:
   - `steps_per_phase=5000`
   - `save_freq=1000`
2. Wait until a checkpoint is saved.
3. Stop the kernel manually.
4. Restart the kernel.
5. Run the resume cell.
6. Verify that the loaded values match the values recorded before stopping.


```
# Fill these values

RUN_NAME = "test_run"
SEED = 42
RESUME_DIR = f"artifacts/walker/{RUN_NAME}/seed_{SEED}"
```

```
from pathlib import Path
import torch

ckpt = Path(RESUME_DIR) / "checkpoint_sf.pt"

if ckpt.exists():
    checkpoint = torch.load(ckpt, map_location="cpu")

    print("=== CHECKPOINT CONTENTS ===")
    print("Phase:", checkpoint["phase"])
    print("Step:", checkpoint["step"])
    print("Buffer size:", len(checkpoint["replay_buffer_storage"]))
    print("Buffer ptr:", checkpoint["replay_buffer_ptr"])
    print("Episode return:", checkpoint["episode_return"])

else:
    print("Checkpoint not found:", ckpt)
```

## Before stopping the run

Write down:

- Phase
- Step
- Buffer size
- Buffer ptr
- Episode return

Then interrupt execution.


```
# Example resume command

print(
    f'''
python train.py \\
  --run_name {RUN_NAME} \\
  --seed {SEED} \\
  --resume_dir {RESUME_DIR}
'''
)
```

## Expected result

After restarting and loading the checkpoint:

- Same phase
- Same checkpoint step
- Same replay buffer size
- Same replay buffer pointer
- Same episode return

Training should continue without errors.
